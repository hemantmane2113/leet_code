```python
def reverse( x: int) -> int:
    digit = 0
    int_list = []

    # 1.handle 0
    if x == 0:
        return x

    # 2. handle negatives
    is_negative =  x < 0
    x = abs(x)

    while x > 0:
        digit = x % 10
        int_list.append(str(digit))
        x = x // 10

    new_int = ''.join(int_list)
    result = int(new_int)

    # 3. negative sign for negative number
    if is_negative:
            result = -result
            
    # 4. 32-bit signed integer overflow check
    if result < -2147483648 or result > 2147483647:
        return 0
        
    return result
```

```python
# best solution 
def reverse(x: int) -> int:
    # 1. Determine the sign (1 for positive, -1 for negative)
    sign = -1 if x < 0 else 1
    
    # 2. Make it positive, turn to string, reverse it, convert back to int
    reversed_int = int(str(abs(x))[::-1]) * sign
    
    # 3. 32-bit signed integer overflow boundaries
    if reversed_int < -2**31 or reversed_int > 2**31 - 1:
        return 0
        
    return reversed_int

# --- Test Cases ---
print(reverse(123))         # Output: 321
print(reverse(-123))        # Output: -321
print(reverse(120))         # Output: 21 (automatically drops leading zero)
print(reverse(0))           # Output: 0
print(reverse(1534236469))  # Output: 0 (overflows 32-bit limit)
```

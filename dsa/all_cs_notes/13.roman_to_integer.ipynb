```python
"""
 logic for matching individual symbols is correct, but this code will fail on strings 
 containing the subtractive forms (like "IV", "IX", "XL", etc.) 
 because it splits them into separate single letters.

"""
def romanToInt(s: str) -> int:
    # Symbol as KEY, Integer as VALUE (Ordered from LARGEST to SMALLEST)
    symb_dict = {
        "M": 1000, "CM": 900, "D": 500, "CD": 400,
        "C": 100,  "XC": 90,  "L": 50,  "XL": 40,
        "X": 10,   "IX": 9,   "V": 5,   "IV": 4,
        "I": 1
    }
    s_list = list(s)
    
    result = 0
    
    # Loop unpacks key (symbol) and value (integer) top-to-bottom
    for i in s_list:
        result += symb_dict[i]
             
    return result
```

```python
def romanToInt(s: str) -> int:
    # We only need the base single-character symbols for this direction
    symb_dict = {
        "M": 1000,
        "D": 500,
        "C": 100,
        "L": 50,
        "X": 10,
        "V": 5,
        "I": 1
    }
    
    result = 0
    
    # Loop through the string by index
    for i in range(len(s)):
        # Look ahead: If this value is smaller than the next value, subtract it
        if i + 1 < len(s) and symb_dict[s[i]] < symb_dict[s[i+1]]:
            result -= symb_dict[s[i]]
        # Otherwise, add it normally
        else:
            result += symb_dict[s[i]]
             
    return result

# --- Test Cases ---
print(romanToInt("LVIII"))   # Output: 58   (50 + 5 + 1 + 1 + 1)
print(romanToInt("MCMXCIV")) # Output: 1994 (1000 - 100 + 1000 - 10 + 100 - 1 + 5)
print(romanToInt("IV"))      # Output: 4    (-1 + 5)
```

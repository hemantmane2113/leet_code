```python
#works
# we have modified the solution to return the new sorted list with only unique elements as well
class Solution:
    def removeDuplicates(self, nums: list[int]) -> tuple[int, list[int]]:
        if not nums:
            return 0, []

        index = 1

        for i in range(1, len(nums)):
            if nums[i] != nums[i-1]:
                nums[index] = nums[i]
                index += 1
        
        # Slice the modified array to get only the unique elements
        unique_list = nums[:index]
        
        # Return both the count (index) and the new list as a tuple
        return index, unique_list
```

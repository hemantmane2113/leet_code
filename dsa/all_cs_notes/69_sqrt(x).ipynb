```python
def binarysearch(nums_list,target):
    if not nums_list:
        return -1

    if nums_list[0] ==  target:
        return 0

    low = 0
    high = len(nums_list) - 1

    while low <= high:
        mid = low + (high - low) // 2

        if nums_list[mid] ==  target:
            return mid
        elif mid < target:
            low = mid + 1
        
        else:
            high  = mid - 1

    
    return -1
```

```python
nums_list = [12,20,45,56,67,88,91]
target = 58
iret = binarysearch(nums_list,target)
print(iret)
```

```python
def mySqrt(self, x: int) -> int:
    if x < 2:
        return x

    left =  1
    right = x // 2

    while left <= right:
        mid = (left + right) // 2
        mid_squared = mid * mid

        if mid_squared == x:
            return mid
        elif mid_squared < x:
            left = mid + 1
        else:
            right = mid - 1

    return right
```

```python
x = 36
iret = mySqrt(None,36)
print(iret)
```

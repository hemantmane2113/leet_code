```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def addTwoNumbers(self, l1: ListNode | None, l2: ListNode | None) -> ListNode | None:
        temp_node = ListNode(0)
        current = temp_node
        carry = 0

        p = l1
        q = l2

        while p is not None or q is not None or carry:
            digit1 = p.val if p is not None else 0
            digit2 = q.val if q is not None else 0

            total = digit1 + digit2 + carry

            digit = total % 10
            carry = total // 10

            current.next = ListNode(digit)
            current = current.next

            if p is not None:
                p = p.next

            if q is not None:
                q = q.next

        return temp_node.next
        
```

```python
l1 = [2,4,3]
l2 = [5,6,4]

sol = Solution()

iret = sol.addTwoNumbers(l1,l2)

print(iret)

# 5. Helper to print the result cleanly
def print_linked_list(node):
    res = []
    while node:
        res.append(str(node.val))
        node = node.next
    print(" -> ".join(res))

print_linked_list(iret)  # Outputs: 7 -> 0 -> 8
```

Linked List Pattern: Add Two Numbers --- Step-by-Step Workflow

This note is based on my own implementation of LeetCode 2 --- Add
Two Numbers.

1. My ListNode Class

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

Each node has two attributes:

val  → value stored in the node
next → reference to the next node

For example:

node = ListNode(5)

creates:

node
 ↓
[ 5 | None ]

Why can ListNode(0) take only one argument?

The constructor is:

def __init__(self, val=0, next=None):

self is supplied automatically by Python.

val has a default value of 0.

next has a default value of None.

Therefore:

ListNode()

means:

val = 0
next = None

and:

ListNode(0)

explicitly supplies val = 0.

Both produce:

[ 0 | None ]

2. Understand the Input

Example:

l1 = 2 → 4 → 3
l2 = 5 → 6 → 4

The digits are stored in reverse order.

Therefore:

l1 represents 342
l2 represents 465

So:

342 + 465 = 807

The required linked-list result is:

7 → 0 → 8

The reverse order is actually helpful because normal addition starts
from the rightmost digit.

3. Create the Dummy Node

My code:

temp_node = ListNode(0)

This creates an extra node:

temp_node
    ↓
[ 0 | None ]

This node is called a dummy node or sentinel node.

It is not part of the final answer.

I could also write:

dummy = ListNode(0)

dummy is simply a clearer conventional name.

4. Why Do I Need a Dummy Node?

Suppose the answer is:

7 → 0 → 8

Without a dummy node, I would have to create the first node separately:

head = ListNode(7)
current = head

Then create the remaining nodes.

With the dummy node, I can use exactly the same operation for every
result digit:

current.next = ListNode(digit)
current = current.next

The structure becomes:

temp_node → 7 → 0 → 8

The first 0 belongs to temp_node, not to the answer.

Therefore the actual answer starts at:

temp_node.next

5. current = temp_node

My code:

current = temp_node

Initially:

temp_node
    ↓
[ 0 | None ]
    ↑
 current

Both variables refer to the same node.

Important:

temp_node stays at the beginning.

current moves as the result list is constructed.

Think:

temp_node  → fixed reference
current    → moving reference

6. Initialize Carry

carry = 0

Initially there is no carry.

Normal addition sometimes produces a carry:

6 + 4 = 10

So:

digit = 0
carry = 1

The carry is added to the next pair of digits.

7. Create Pointers for the Input Lists

My code:

p = l1
q = l2

Now:

p
↓
2 → 4 → 3


q
↓
5 → 6 → 4

p moves through l1.

q moves through l2.

After processing a node:

p = p.next
q = q.next

8. The Main Loop

My code:

while p is not None or q is not None or carry:

The loop continues when at least one of these is true:

p still exists
OR
q still exists
OR
carry exists

This is important because:

The lists can have different lengths.

A carry can remain after both lists are exhausted.

Example:

99 + 1 = 100

The final 1 comes from the carry.

9. Get the First Digit

digit1 = p.val if p is not None else 0

This means:

If p exists, take its value. Otherwise use 0.

Equivalent code:

if p is not None:
    digit1 = p.val
else:
    digit1 = 0

Why?

Because the lists may have different lengths.

Example:

l1 = 9 → 9
l2 = 1

Eventually q becomes None.

Instead of trying:

q.val

we treat the missing digit as:

0

10. Get the Second Digit

digit2 = q.val if q is not None else 0

Same concept.

If q exists:

digit2 = q.val

Otherwise:

digit2 = 0

11. Add the Digits

total = digit1 + digit2 + carry

This represents normal addition.

For example:

digit1 = 4
digit2 = 6
carry  = 0

total = 4 + 6 + 0
      = 10

12. Extract the Digit

digit = total % 10

Every result node must contain one digit.

For:

total = 7

we get:

7 % 10 = 7

For:

total = 10

we get:

10 % 10 = 0

For:

total = 18

we get:

18 % 10 = 8

Therefore % 10 gives us the digit that belongs in the current result
node.

13. Calculate the Carry

carry = total // 10

Integer division gives the carry.

Examples:

7 // 10  = 0
10 // 10 = 1
18 // 10 = 1

Therefore:

total = 18

digit = 8
carry = 1

This is equivalent to normal manual addition.

14. Create a Result Node

My code:

current.next = ListNode(digit)

Suppose:

digit = 7

Initially:

temp_node → 0 → None
              ↑
            current

After:

current.next = ListNode(7)

we have:

temp_node → 0 → 7 → None
              ↑
            current

The new node has been attached to the result list.

15. Move current

My code:

current = current.next

Before:

temp_node → 0 → 7 → None
              ↑
            current

After:

temp_node → 0 → 7 → None
                  ↑
                current

Now current points to the newly created last node.

This allows the next digit to be attached after it.

16. Move p

My code:

if p is not None:
    p = p.next

Suppose:

p
↓
2 → 4 → 3

After:

p = p.next

we have:

    p
    ↓
2 → 4 → 3

p now points to the next node containing 4.

17. Move q

My code:

if q is not None:
    q = q.next

Same idea.

q
↓
5 → 6 → 4

becomes:

    q
    ↓
5 → 6 → 4

q now points to 6.

We constructed:

21. The Algorithm in Plain English

When I see this problem, I should think:

Start at both linked lists
        ↓
Create a dummy node for the answer
        ↓
Set carry = 0
        ↓
Read one digit from each list
        ↓
If a list has ended, use 0
        ↓
Add digit1 + digit2 + carry
        ↓
Extract the result digit using % 10
        ↓
Extract carry using // 10
        ↓
Create a new result node
        ↓
Move current
        ↓
Move p and q
        ↓
Repeat while a node or carry remains
        ↓
Return dummy.next

22. Complexity

Let n be the length of the longer linked list.

Time Complexity

Every input node is processed once:

O(n)

Output Space

The result contains up to n + 1 nodes:

O(n)

Auxiliary Space

Apart from the output list, the algorithm uses only a constant number of
variables:

O(1)

23. Reusable Linked-List Workflow

When I don't know how to approach a linked-list problem, use this
checklist.

Step 1 --- Understand the nodes

Ask:

What does each node represent?

Step 2 --- Identify the pointers

Ask:

Which nodes do I need to keep track of?

Common pointers:

current
previous
next_node
slow
fast
p
q

Step 3 --- Identify extra state

Ask:

Do I need something besides pointers?

Examples:

carry

previous value

Step 4 --- Determine movement

Usually:

Step 5 --- Determine the operation

Ask:

What should happen to the current node?

Examples:

add values
delete node
reverse link

compare nodes

Step 6 --- Check edge cases

Think about:

empty list
one node
different lengths
end of list
remaining carry

Step 7 --- Determine the stopping condition

Ask:

When do I know the work is finished?

Step 8 --- Determine what to return

Ask:

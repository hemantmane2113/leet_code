```python
def climbStairs(self, n: int) -> int:
    if n == 1:
        return 1
    if n == 2:
        return 2
    one_step_at_a_time = 1
    two_steps_at_a_time = 2
    total_ways = 0

    for i in range(3,n+1):
        total_ways = one_step_at_a_time + two_steps_at_a_time
        one_step_at_a_time = two_steps_at_a_time
        two_steps_at_a_time = total_ways

    return total_ways
```

```python
def showStairsPaths(n: int):
    results = []

    def find_paths(remaining_steps, current_path):
        # Base Case: If no steps are left, we successfully reached the top!
        if remaining_steps == 0:
            results.append(list(current_path))
            return
        
        # Scenario 1: Try taking a single step (1) if enough steps remain
        if remaining_steps >= 1:
            current_path.append(1)
            find_paths(remaining_steps - 1, current_path)
            current_path.pop() # Backtrack: remove the 1 to try a different move
            
        # Scenario 2: Try taking a double step (2) if enough steps remain
        if remaining_steps >= 2:
            current_path.append(2)
            find_paths(remaining_steps - 2, current_path)
            current_path.pop() # Backtrack: remove the 2

    # Start the calculation with 'n' steps remaining and an empty path
    find_paths(n, [])
    
    # Print out all the paths beautifully
    print(f"--- All distinct ways to climb {n} stairs ---")
    for path in results:
        print(" -> ".join(map(str, path)))

# Let's run it for 4 stairs!
showStairsPaths(10)
```

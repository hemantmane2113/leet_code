```python
# works
def threeSumClosest(nums,target):
    n = len(nums)

    sorted_nums = sorted(nums)

    closest_sum = float('inf')

    for i in range(n):
        left = i + 1
        right = n - 1

        while left < right:
            current_sum = sorted_nums[i] + sorted_nums[left] + sorted_nums[right]

            if abs(current_sum - target) < abs(closest_sum - target):
                closest_sum = current_sum
            if current_sum < target:
                left += 1
            elif current_sum > target:
                right -= 1
            else:
                return current_sum
            
    return closest_sum
```

```python
# works but little optimised
def threeSumClosest(nums,target):
    n = len(nums)

    sorted_nums = sorted(nums)

    closest_sum = float('inf')

    for i in range(n -2):
        # we need 3 elements and by the time i reaches the last but third element,left will be at last but second and
        # right will be at last element,so there is no point in taking i till the last element
        left = i + 1
        right = n - 1

        while left < right:
            current_sum = sorted_nums[i] + sorted_nums[left] + sorted_nums[right]

            if abs(current_sum - target) < abs(closest_sum - target):
                closest_sum = current_sum
            if current_sum < target:
                left += 1
            elif current_sum > target:
                right -= 1
            else:
                return current_sum
            
    return closest_sum
```


Time Complexity: O(n^2)

• Sorting the array takes O(n \log n) time.
• The outer loop runs n - 2 times, which simplifies to O(n).
• The inner two-pointer loop (while left < right) processes the remaining elements linearly, taking O(n) time per iteration of the outer loop.
• Multiplying the loops gives O(n \times n) = O(n^2). Since O(n^2) dominates O(n \log n), the total time complexity is O(n^2).

Space Complexity: O(1) to O(n)

• The space complexity depends entirely on the programming language's implementation of the .sort() method.
• In Python, .sort() uses the Timsort algorithm, which requires O(n) auxiliary space in the worst case to store temporary arrays.
• Aside from the memory used by the sorting algorithm, the code only tracks a few pointers and variables (left, right, closest_sum), consuming O(1) auxiliary memory.

1. What is closest_sum = float('inf')?

At the very start, we haven't checked any numbers yet. We initialize closest_sum to infinity (float('inf')).
• We do this because infinity is the worst possible "closest sum."
• Any real number we calculate later will easily beat infinity and take its place.

2. Breaking Down the if Statement

The line if abs(current_sum - target) < abs(closest_sum - target): is simply asking:
"Is our new combination closer to the target than our old best combination?"
Let's look at the two sides of that < sign using an example:
• Say your target is 10.
• Your current scoreboard (closest_sum) is holding 4. (Distance to target: 10 - 4 = 6)
• You just calculated a new triplet (current_sum) that equals 11. (Distance to target: 11 - 10 = 1)
Now let's plug those numbers into the if statement:

Left Side: abs(current_sum - target)

• This calculates how far away the new sum is from the target.
• abs(11 - 10) --> abs(1) --> 1
• (The abs() function turns negative distances positive, because being 2 steps below the target is just as far away as being 2 steps above it).

Right Side: abs(closest_sum - target)

• This calculates how far away our previous best record was from the target.
• abs(4 - 10) --> abs(-6) --> 6

The Comparison:

• The line becomes: if 1 < 6:
• Since 1 is smaller than 6, it means 11 is much closer to the target than 4.
• The if statement evaluates to True, and the code updates the scoreboard: closest_sum = current_sum (so closest_sum becomes 11).

```python
# more optimised solution
# no change time and space complexity is O(n^2) and O(1) respectively

def threeSumClosest(nums, target):
    nums.sort()
    closest_sum = float('inf')
    n = len(nums)
    
    for i in range(n - 2):
        # Optimization 1: Skip duplicate values for the first element
        if i > 0 and nums[i] == nums[i - 1]:
            continue
            
        left = i + 1
        right = n - 1
        
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]
            
            if current_sum == target:
                return current_sum  # Ultimate shortcut: exact match found
                
            if abs(current_sum - target) < abs(closest_sum - target):
                closest_sum = current_sum
                
            if current_sum < target:
                left += 1
                # Optimization 2: Skip duplicate values for the left pointer
                while left < right and nums[left] == nums[left - 1]:
                    left += 1
            else:
                right -= 1
                # Optimization 3: Skip duplicate values for the right pointer
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1
                    
    return closest_sum
```

```python
# works
from itertools import product

def letterCombinations(self, digits: str) -> list[str]:
    # Edge case: if input is empty, return an empty list
    if not digits:
        return []
        
    phone_dict = {
        '2': "abc", '3': 'def', '4': "ghi", '5': "jkl",
        '6': "mno", '7': "pqrs", '8': 'tuv', '9': 'wxyz'
    }
    
    # 1. Map each input digit to its corresponding string of letters
    # Example for "23": ["abc", "def"]
    strings = [phone_dict[d] for d in digits]
    
    # 2. Use itertools.product with * to find all cross-group combinations
    # 3. Join each combination tuple into a single string
    return ["".join(comb) for comb in product(*strings)]
```

```python
digits = '23'

iret = letterCombinations(None, digits)

print(iret)

```

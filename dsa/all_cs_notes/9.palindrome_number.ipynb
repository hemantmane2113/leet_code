```python
# works
# time complexity:O(N)
# space complexity: O(N)

def isPalindrome(x: int) -> bool:
  
    if x < 0:
        return False
    
    x = str(x)
    n = len(x)

    left = 0
    right = n - 1

    while left <= right:
        if x[left] == x[right]:
            left += 1
            right -= 1
        else:
            return False
    return True
```

```python
# The Math Approach (No Strings Allowed)
# time complexity:O(N)
# space complexity: O(1)
def isPalindrome(x: int) -> bool:
    # Negative numbers and numbers ending in 0 (except 0 itself) are not palindromes
    if x < 0 or (x % 10 == 0 and x != 0):
        return False
    
    original = x
    reversed_num = 0
    
    while x > 0:
        # Get the last digit (e.g., 123 % 10 = 3)
        last_digit = x % 10
        # Shift existing digits left and add the new one (e.g., 12 * 10 + 3 = 123)
        reversed_num = (reversed_num * 10) + last_digit
        # Remove the last digit from x (e.g., 123 // 10 = 12)
        x = x // 10
        
    return original == reversed_num
```

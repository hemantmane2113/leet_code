```python
#time complexity : O(1)
#space complexity : O(1)
def intToRoman(num: int) -> str:
    # Integer as KEY and symbol as VALUE(ordered from largest to smallest)
    symb_dict = {
        1000: "M", 900: "CM", 500: "D", 400: "CD",
        100: "C",  90: "XC",  50: "L",  40: "XL",
        10: "X",   9: "IX",   5: "V",   4: "IV",
        1: "I"
    }

    result = []

    # Iterate through from largest math value down to 1
    for value, key in symb_dict.items():
        while num >= value:
            result.append(key)
            num -= value
                
    return "".join(result)
```

```python
# --- Test Cases ---
print(intToRoman(3742)) # Output: "MMMDCCXLII" (3000=MMM, 700=DCC, 40=XL, 2=II)
print(intToRoman(58))   # Output: "LVIII"      (50=L, 8=VIII)
print(intToRoman(1994)) # Output: "MCMXCIV"    (1000=M, 900=CM, 90=XC, 4=IV)
print(intToRoman(2000)) # Output: "MM"       (While loop runs twice consecutively for "M")
print(intToRoman(1994)) # Output: "MCMXCIV"  (Processes each place value accurately)
print(intToRoman(58))   # Output: "LVIII"    (Skips values until it hits 50)
```

```python
def intToRoman(num: int) -> str:
    # Symbol as KEY, Integer as VALUE (Ordered from LARGEST to SMALLEST)
    symb_dict = {
        "M": 1000, "CM": 900, "D": 500, "CD": 400,
        "C": 100,  "XC": 90,  "L": 50,  "XL": 40,
        "X": 10,   "IX": 9,   "V": 5,   "IV": 4,
        "I": 1
    }
    
    result = []
    
    # Loop unpacks key (symbol) and value (integer) top-to-bottom
    for symbol, value in symb_dict.items():
        while num >= value:
            result.append(symbol)
            num -= value
             
    return "".join(result)
```

```python
# --- Test Cases ---
print(intToRoman(3742)) # Output: "MMMDCCXLII" (3000=MMM, 700=DCC, 40=XL, 2=II)
print(intToRoman(58))   # Output: "LVIII"      (50=L, 8=VIII)
print(intToRoman(1994)) # Output: "MCMXCIV"    (1000=M, 900=CM, 90=XC, 4=IV)
print(intToRoman(2000)) # Output: "MM"       (While loop runs twice consecutively for "M")
print(intToRoman(1994)) # Output: "MCMXCIV"  (Processes each place value accurately)
print(intToRoman(58))   # Output: "LVIII"    (Skips values until it hits 50)
```

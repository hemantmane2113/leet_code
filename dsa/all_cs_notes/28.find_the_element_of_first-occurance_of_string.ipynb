```python
#works
# without using built-in functions
def strStr(self, haystack: str, needle: str) -> int:
    # Get the lengths of both strings
    h_len = len(haystack)
    n_len = len(needle)
    
    # Loop through the haystack, stopping where the needle can no longer fit
    for i in range(h_len - n_len + 1):
        # Slice a window of the same length as the needle and compare
        if haystack[i : i + n_len] == needle:
            return i  # Return the starting index immediately
            
    return -1  # If no match is found after checking all windows
```

 Case 1: If there is a match
Step 1: Setting up the boundaries

Before the loop starts, Python counts the lengths:
• h_len = 9
• n_len = 3
• The loop range math is: 9 - 3 + 1 = 7. This means i will check indices 0, 1, 2, 3, 4, 5, and 6.

Step 2: The Loop Starts (At index i = 0)

The computer sets i = 0. Now it evaluates the if condition:
1. Calculate the window coordinates:
i : i + n_len becomes 0 : 0 + 3, which is 0 : 3.
2. Slice the haystack:
haystack[0:3] means "give me characters at index 0, 1, and 2".
Looking at "sadbutsad":
	• Index 0 = s
	• Index 1 = a
	• Index 2 = d
So, haystack[0:3] extracts the string "sad".
3. Compare with the needle:
The code checks: Is "sad" == "sad"?
Yes! It is a perfect match.
4. Return the result:
Because it is a match, the code immediately runs return i. Since i is currently 0, it returns 0 and the entire function stops.

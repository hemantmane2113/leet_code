```python
# works
def plusOne(self, digits: list[int]) -> list[int]:
    n = len(digits)
    carry = 1

    for i in range(n-1, -1, -1):
        # Add the carry to the current digit
        total = digits[i] + carry
        if total < 10:
            digits[i] = total
            carry = 0  # Carry is spent
            break      # No more work needed
        else:
            digits[i] = 0
            carry = 1  # Carry moves to the next left digit

    # If carry is still 1 after the loop, it means we had an overflow (e.g., 99 -> 100)
    if carry == 1:
        digits.insert(0, 1)
    
    return digits
```

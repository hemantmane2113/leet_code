```python
# complexity
#Time  : O(n²)
#Space : O(1)

def longestPalindrome(s: str) -> str:
    n = len(s)

    if n <= 1:
        return s

    max_length = 1
    start = 0
    end = 0

    def expand(left, right):

        while left >= 0 and right < n and s[left] == s[right]:
            left -= 1
            right += 1

        # left and right have gone one step too far
        return left + 1, right - 1

    for center in range(n):

        # Odd-length palindrome
        left1, right1 = expand(center, center)

        length1 = right1 - left1 + 1

        if length1 > max_length:
            max_length = length1
            start = left1
            end = right1

        # Even-length palindrome
        left2, right2 = expand(center, center + 1)

        length2 = right2 - left2 + 1

        if length2 > max_length:
            max_length = length2
            start = left2
            end = right2

    return s[start:end + 1]
```

```python
# best version

# complexity
#Time  : O(n²)
#Space : O(1)
def longestPalindrome(s: str) -> str:

    n = len(s)

    if n <= 1:
        return s

    start = 0
    max_length = 1

    def expand(left, right):

        while (
            left >= 0
            and right < n
            and s[left] == s[right]
        ):
            left -= 1
            right += 1

        return right - left - 1

    for center in range(n):

        # Odd-length palindrome
        odd_length = expand(center, center)

        # Even-length palindrome
        even_length = expand(center, center + 1)

        current_length = max(odd_length, even_length)

        if current_length > max_length:

            max_length = current_length

            # Calculate starting index
            start = center - (current_length - 1) // 2

    return s[start:start + max_length]
```

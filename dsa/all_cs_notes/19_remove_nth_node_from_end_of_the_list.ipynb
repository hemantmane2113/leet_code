```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def removeNthFromEnd(self, head: ListNode, n: int) -> ListNode:
        # 1. Create a dummy node pointing to the head
        dummy = ListNode(0)
        dummy.next = head
        
        fast = dummy
        slow = dummy
        
        # 2. Advance the fast pointer by n + 1 steps
        for _ in range(n + 1):
            fast = fast.next
            
        # 3. Move both pointers together until fast reaches the end
        while fast is not None:
            fast = fast.next
            slow = slow.next
            
        # 4. Remove the Nth node from the end
        slow.next = slow.next.next
        
        # 5. Return the real head
        return dummy.next
```

```python
# works but time complexity is O(n) and required is O(logn)
def searchRange(self, nums: list[int], target: int) -> list[int]:
    if not nums:
        return [-1,-1]

    n = len(nums)
    result = []

    i = 0
    j = n - 1

    for i in range(n):
        if nums[i] == target:
            result.append(i)
            break
    
    for j in range(n-1,-1,-1):
        if nums[j] == target:
            result.append(j)
            break

    if not result:
        result = [-1,-1]

    return result
```

```python
# works
# time complexity: O(logn)
# but uses inbuilt function
import bisect

class Solution:
    def searchRange(self, nums: list[int], target: int) -> list[int]:
        # Find the insertion point for target (first occurrence)
        left = bisect.bisect_left(nums, target)
        
        # If target isn't in the list or left index is out of bounds
        if left == len(nums) or nums[left] != target:
            return [-1, -1]
            
        # Find the insertion point for target from the right (one past last occurrence)
        right = bisect.bisect_right(nums, target) - 1
        
        return [left, right]
```

```python
# works
# time complexity:O(logn)
def searchRange(self, nums: list[int], target: int) -> list[int]:
    def find_bound(is_first: bool) -> int:
        left ,right = 0, len(nums) - 1
        bound = -1
        
        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                bound = mid
                if is_first:
                    right = mid - 1  # Keep looking to the left
                else:
                    left = mid + 1   # Keep looking to the right
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
                
        return bound

    return [find_bound(True), find_bound(False)]
```

```python
# works,simplified version of above code
# time complexity:O(logn)
def searchRange(self, nums: list[int], target: int) -> list[int]:

    # Find first position
    left = 0
    right = len(nums) - 1
    first = -1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            first = mid
            right = mid - 1
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    # Find last position
    left = 0
    right = len(nums) - 1
    last = -1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            last = mid
            left = mid + 1
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return [first, last]
```

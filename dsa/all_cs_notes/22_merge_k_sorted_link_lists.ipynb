```python
# Approach 1
# works
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def mergeKLists(self, lists: list[ListNode | None]) -> ListNode | None:
        # Handle the base case where the input list is empty
        if not lists:
            return None

        def merge_two_lists(list1: ListNode | None, list2: ListNode | None) -> ListNode | None:
            dummy = ListNode(-1)
            temp = dummy

            # Traverse both lists using the list variables directly
            while list1 is not None and list2 is not None:
                if list1.val <= list2.val:  # Fixed: Compare values, not nodes
                    temp.next = list1
                    list1 = list1.next      # Fixed: Advance the actual list pointer
                else:
                    temp.next = list2
                    list2 = list2.next      # Fixed: Advance the actual list pointer
                temp = temp.next

            # Attach remaining elements
            if list1:
                temp.next = list1
            else:
                temp.next = list2

            return dummy.next  # Fixed: Return dummy.next, not temp.next

        # Iteratively merge all lists into the first one
        result = lists[0]
        for i in range(1, len(lists)):
            result = merge_two_lists(result, lists[i])
            
        return result  # Fixed: Missing final return statement
```

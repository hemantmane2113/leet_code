```python
#brute force
def twoSum(nums: list[int], target: int) -> list[int]:
    n = len(nums)
 
    for i in range(n):
        for j in range(i+1,n):
            if target == nums[i] + nums[j]:
                return [i,j]
    return []
```

```python
# hash map
def twoSum(nums: list[int], target: int) -> list[int]:
    n = len(nums)
    seen = {}

    for i,num in enumerate(nums):
        complementary = target - num
        if complementary in seen:
            return [seen[complementary],i]
        seen[num] = i
    return []
```

```python
# nums = [2,7,11,15]
# target = 9
# nums = [3,2,4]
# target = 6
nums = [3,3]
target = 6
iret = twoSum(nums,target)
print(iret)
```

```python
# binary search algorithm
def searchInsert(self, nums: list[int], target: int) -> int:
    low = 0
    high = len(nums) - 1

    while low <= high:
        mid = low + (high -low)//2

        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            low = mid + 1
        else:
            high = mid -1
    
    return -1
```

```python
# works
# time complexity: O(logn)
def searchInsert(self, nums: list[int], target: int) -> int:
    low = 0
    high = len(nums) - 1

    while low <= high:
        mid = low + (high -low)//2

        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            low = mid + 1
        else:
            high = mid -1
    
    return low
    
```

"low" represents the first position where target can be inserted without breaking the sorted order,when the binary search fails to find the target.

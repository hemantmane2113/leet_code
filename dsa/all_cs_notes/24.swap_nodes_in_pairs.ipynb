
The  Bugs in the Code

1. AttributeError on Empty or Single-Node Lists: If head is empty (None) or only has one node, trying to access 
q = p.next or current.next will immediately throw an AttributeError: 'NoneType' object has no attribute 'next'.

2. Crash on Odd-Length Lists: In your loop condition while current.next is not None:, if your list has an odd number of elements, current.next.next will eventually become None. On the next iteration, checking current.next will crash because current is None.

3. Values Overwritten Incorrectly: Inside the loop, you have:

p.val = q.val  # Overwrites p.val with q.val
q.val = current.val  # But current is the exact same node as p! So you copy the new value back.

Because p and current point to the same node, your original p.val is lost before you can copy it to q. You need a standard Python swap logic: p.val, q.val = q.val, p.val.

```python
#Approach1: Modify/Exchange Nodes
# works but fail to satisfy the condition
# You are not allowed to modify or exchange node values (p.val, q.val = q.val, p.val).
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def swapPairs(self, head: ListNode | None) -> ListNode | None:
        # Fix 1: Guard against empty lists or lists with only 1 node
        if not head or not head.next:
            return head

        current = head
        p = current
        q = p.next

        # Fix 2: Check both current and current.next to safely handle odd lengths
        while current is not None and current.next is not None:
            # Fix 3: Use a safe simultaneous swap so values aren't lost
            p.val, q.val = q.val, p.val
            
            # Safely move current forward by two steps
            current = current.next.next
            
            # Re-align your tracking pointers if we haven't reached the end
            if current and current.next:
                p = current
                q = p.next

        return head
```

```python
#Approach 2:The Iterative Approach (Using a Dummy Node) — Best Overall
class Solution:
    def swapPairs(self, head: ListNode | None) -> ListNode | None:
        # Create a dummy node that points to the head
        dummy = ListNode(-1)
        dummy.next = head
        
        # 'prev' tracks the node right before the pair we want to swap
        prev = dummy
        
        # Ensure there are at least two nodes left to swap
        while prev.next and prev.next.next:
            # Identify the two nodes to be swapped
            first = prev.next
            second = prev.next.next
            
            # Rearrange the pointers
            first.next = second.next  # 1. Connect first node to the rest of the list
            second.next = first       # 2. Turn second node around to point to first
            prev.next = second        # 3. Connect the previous part to the new pair head
            
            # Move prev forward by two nodes for the next pair loop
            prev = first
            
        return dummy.next
```

```python
#Works
#Approach3:The Recursive Approach — Cleanest Code
class Solution:
    def swapPairs(self, head: ListNode | None) -> ListNode | None:
        # Base case: If 0 or 1 node left, no swapping needed
        if not head or not head.next:
            # Return the node as it is
            return head
        
        # Nodes to be swapped
        first = head
        second = head.next
        
        # Recursive step: 
        # first.next should point to the head of the next swapped pair
        first.next = self.swapPairs(second.next)
        
        # Put second before first
        second.next = first
        
        # 'second' is now the new head of this pair
        return second
```


Approach	                   Time Complexity	Space Complexity	    Real-World Performance
Value Swapping (Your Idea)	       {O}(N)	        {O}(1)              Bad if node contains heavy objects
Iterative Pointer Swapping	       {O}(N)	        {O}(1)	            Optimal (Fastest & memory-efficient)
Recursive Pointer Swapping	       {O}(N)	        {O}(N)	            Cleanest code, but uses memory stack

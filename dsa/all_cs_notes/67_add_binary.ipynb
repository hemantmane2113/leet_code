```python
# works
def add_binary(bin1, bin2):
    result = []
    carry = 0
    
    # Pointers starting at the rightmost end of both strings
    i = len(bin1) - 1
    j = len(bin2) - 1
    
    # Loop as long as there are digits to process or a carry remains
    while i >= 0 or j >= 0 or carry:
        total = carry
        
        # Add the digit from the first string if available
        if i >= 0:
            if bin1[i] == '1':
                total += 1
            else:
                total = 0
            i -= 1
            
        # Add the digit from the second string if available
        if j >= 0:
            if bin2[j] == '1' :
                total += 1
            else:
                total = 0
            j -= 1
            
        # Determine the output digit and the new carry
        # If total is 0 -> digit 0, carry 0
        # If total is 1 -> digit 1, carry 0
        # If total is 2 -> digit 0, carry 1
        # If total is 3 -> digit 1, carry 1
        result.append(str(total % 2))
        carry = total // 2
        
    # Since digits were added from right to left, reverse the result
    return "".join(result[::-1])
```

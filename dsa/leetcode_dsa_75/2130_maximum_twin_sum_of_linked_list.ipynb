```python
# best solution:internet
# time complexity: O(n)
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def pairSum(self, head: Optional[ListNode]) -> int:
        # 1. Find the middle of the linked list
        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # 2. Reverse the second half
        previous = None

        while slow:
            ahead = slow.next
            slow.next = previous
            previous = slow
            slow = ahead

        # 3. Compare first half with reversed second half
        current = head
        max_sum = 0

        while previous:
            max_sum = max(max_sum, current.val + previous.val)
            current = current.next
            previous = previous.next

        return max_sum
```

```python
# 1.Helper function to turn a Python list into Linked List nodes
def make_linked_list(arr):
    if not arr: return None
    
    head = ListNode(arr[0])
    current = head
    for num in arr[1:]:
        current.next = ListNode(num)
        current = current.next
    return head
```

```python
# 2.Helper function Convert Linked List back to Python list for easy printing
def print_linked_list(head):
    result = []
    current = head
    while current:
        result.append(str(current.val))
        current = current.next
    print(" -> ".join(result))
```

```python
# works well
#time_complexity: O(n)
def uniqueOccurrences(arr) -> bool:

    arr_dict = {}
    values_list = []

    for element in arr:
        if element in arr_dict:
            arr_dict[element] += 1
        else:
            arr_dict[element] = 1

    for values in arr_dict.values():
        values_list.append(values)

    if len(set(values_list)) == len(values_list):
        return True
    else:
        return False
```

```python
#best solution
# time_complexity: O(n)
def uniqueOccurrences(arr) -> bool:

    arr_dict = {}

    for element in arr:
        if element in arr_dict:
            arr_dict[element] += 1
        else:
            arr_dict[element] = 1

    values_list = list(arr_dict.values())

    return len(set(values_list)) == len(values_list)
```

```python
#arr = [1,2,2,1,1,3]
#arr = [1,2]
arr = [-3,0,1,-3,1,1,1,-3,10,0]
iret = uniqueOccurrences(arr)
print(iret)
```

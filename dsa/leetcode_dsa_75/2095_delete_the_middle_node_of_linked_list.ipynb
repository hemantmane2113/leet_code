1. Using self.length

Here, the linked-list object remembers its length.
time complexty : O(1)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class MySinglyLinkedList:
    def __init__(self):
        self.head = None
        self.length = 0

    # def Length(self):
    #     return self.length...we don't need a seprate function for length calculation

    def IsEmpty(self):
        return self.head is None

    def Display(self):
        if self.IsEmpty():
            print("linked list is empty")
            return

        current = self.head

        while current is not None:
            print(f"|{current.data}|-> ", end="")
            current = current.next

        print()

    def Insert(self, data):
        new = Node(data)

        new.next = self.head
        self.head = new

        self.length += 1  # the length is caculated here

        self.Display()
        print("Length:", self.length)

    def DeleteMiddle(self):
        if self.length == 0:
            print("Invalid length to delete")
            return

        if self.length == 1:
            self.head = None
            self.length -= 1
            self.Display()
            print("Length:", self.length)
            return

        pos = self.length // 2

        p = self.head
        q = None

        for i in range(pos):
            q = p
            p = p.next

        q.next = p.next
        p.next = None

        self.length -= 1  # the length is caculated here


        self.Display()
        print("Length:", self.length)
```

Here, we don't store the length permanently.

time complexity = O(n)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class MySinglyLinkedList:
    def __init__(self):
        self.head = None

    def Length(self):
        current = self.head
        length = 0

        while current is not None:
            length += 1
            current = current.next

        return length

    def IsEmpty(self):
        return self.head is None

    def Display(self):
        if self.IsEmpty():
            print("linked list is empty")
            return

        current = self.head

        while current is not None:
            print(f"|{current.data}|-> ", end="")
            current = current.next

        print()

    def Insert(self, data):
        new = Node(data)

        new.next = self.head
        self.head = new

        self.Display()
        print("Length:", self.Length())

    def DeleteMiddle(self):
        length = self.Length()

        if length == 0:
            print("Invalid length to delete")
            return

        if length == 1:
            self.head = None
            self.Display()
            print("Length:", self.Length())
            return

        pos = length // 2

        p = self.head
        q = None

        for i in range(pos):
            q = p
            p = p.next

        q.next = p.next
        p.next = None

        self.Display()
        print("Length:", self.Length())
```

my solution
time complexity:O(n)
space complexity:O(1)

```python
class Solution:
    def deleteMiddle(self, head: Optional[ListNode]) -> Optional[ListNode]:

        current = head
        length = 0

        # Find length
        while current is not None:
            length += 1
            current = current.next

        # Empty list / one node
        if length <= 1:
            return None

        # Find middle position
        pos = length // 2

        p = head
        q = None

        # Move p to middle and q one node behind it
        for i in range(pos):
            q = p
            p = p.next

        # Delete middle node
        q.next = p.next
        p.next = None

        return head
```

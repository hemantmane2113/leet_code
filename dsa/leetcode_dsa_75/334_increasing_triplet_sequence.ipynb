```python
def increasingTriplet(nums) -> bool:
    n = len(nums)
    left = 0
    middle = 1
    last = 2

    tri1 = nums[left]
    tri2 = nums[middle]
    tri3 = nums[left]

    if n >= 3:
       
        while (left < middle):
            if middle < n and nums[left] < nums[middle] :
                tri1 = nums[left]
                tri2 = nums[middle]
                break

            else:
                if middle >= n:
                    break

            middle += 1
            
                

        while (middle < last):
            if last < n and tri2 < nums[last]:
                tri3 = nums[last]
                break
            else:
                if last >= n:
                    break
            last += 1
                
        if left < middle and middle < last:
            if tri1 < tri2  and tri2 < tri3:
                return True
        
    return False
    
```

```python
def increasingTriplet(nums) -> bool:

    first = float('inf')
    second = float('inf')

    for current in nums:
        if current <= first:
            first = current
        elif current <= second:
            second = current
        else:
            return True

    return False
```

```python
def asteroidCollision(asteroids):
    n = len(asteroids)
    stack = []

    for i in range(n):
        if len(stack) >= 1:
            if asteroids[i] > 0 and stack[-1] > 0:
                stack.append(asteroids[i])
            elif (asteroids[i] < 0 and stack[-1] < 0):
                stack.append(asteroids[i])
            elif (asteroids[i] < 0 and stack[-1] > 0) and (abs(asteroids[i]) >= abs(stack[-1])):
                stack.pop(-1)
            elif asteroids[i] < 0 and stack[-1] > 0 and (abs(asteroids[i]) < abs(stack[-1])):
                stack.append(asteroids[i])
            elif asteroids[i] > 0 and stack[-1] < 0 and (abs(asteroids[i]) > abs(stack[-1])):
                stack.append(asteroids[i])
            elif asteroids[i] > 0 and stack[-1] < 0 and (abs(asteroids[i]) < abs(stack[-1])):
                stack.pop(-1)
            
        else:
            stack.append(asteroids[i])

    return stack
```

```python
def asteroidCollision(asteroids):
    stack = []

    for asteroid in asteroids:
        
        while stack and stack[-1] > 0 and asteroid < 0:
            
            if stack[-1] < abs(asteroid):#eg stack = 10 and asteroid = -12
                stack.pop()              
                continue
            
            elif stack[-1] == abs(asteroid):#eg stack = 10 and asteroid = -10
                stack.pop()
                asteroid = 0
                break
            
            else:#eg stack = 10 and asteroid = -5
                asteroid = 0
                break
        
        if asteroid != 0:
            stack.append(asteroid)

    return stack
```

```python
# best solution,more clean
# time complexity:O(m + n)
# space complexity :O(n)

from collections import Counter

def closeStrings(word1: str, word2: str) -> bool:

    word1_dict = {}
    word2_dict = {}

    for element in word1:
        if element in word1_dict:
            word1_dict[element] += 1
        else:
            word1_dict[element] = 1

    for element in word2:
        if element in word2_dict:
            word2_dict[element] += 1
        else:
            word2_dict[element] = 1

    return (
        set(word1_dict.keys()) == set(word2_dict.keys())
        and
        Counter(word1_dict.values()) == Counter(word2_dict.values())
    )
```

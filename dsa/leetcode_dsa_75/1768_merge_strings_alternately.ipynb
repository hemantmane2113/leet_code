```python
# alternative solution

def mergeAlternately(word1: str, word2: str) -> str:
    result = []

    n1 = len(word1)
    n2 = len(word2)

    n = n1 + n2

    j  = 0
    k = 0

    for i in range(n):
        if (i % 2 == 0) and (j < n1 ):
            result.append(word1[j])
            
            j+= 1
            
            
        elif (i % 2 != 0) and (k < n2 ):
            result.append(word2[k])
          
            k += 1                                                      
        
    result.extend(word1[j:])
    result.extend(word2[k:])

    
    return ''.join(result)
```

```python
str1 = 'pq'
str2 = 'abcd'

bret =mergeAlternately(str1,str2)
print(bret)
```

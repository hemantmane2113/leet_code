```python

def moveZeroes(nums) -> None:

    left = 0

    # Put all non-zero elements at the front
    for right in range(len(nums)):
        if nums[right] != 0:
            nums[left] = nums[right]
            left += 1

    # Fill remaining positions with zero
    while left < len(nums):
        nums[left] = 0
        left += 1
```

```python
#nums = [0,1,0,3,12]
#nums = [0]
nums = [1,0,1]

iret =moveZeroes(nums)
print(iret)
```

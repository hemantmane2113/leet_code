```python
# the solution with time complexity O(n) and without using division operation
def productExceptSelf(nums):

    n = len(nums)

    result1 = [1] * len(nums)
    result2 = [1] * len(nums)


    left_product = 1
    right_product = 1

    for i in range(n):
        result1[i] = left_product
        left_product = left_product * nums[i]

    for i in range(n-1,-1,-1):
        result2[i] =  right_product
        right_product = right_product * nums[i]

    result = [a * b for a, b in zip(result1, result2)]
            

    return result
```

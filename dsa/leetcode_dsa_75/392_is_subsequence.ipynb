```python
# works
def isSubsequence(s,t)->bool:
    left = 0
    right = 0

    while left < len(s) and right < len(t):
        if s[left] != t[right]:
            right += 1
        else:
            left += 1
            right += 1

    if left == len(s):
        return True
    else:
        return False
```

```python
# best code
def isSubsequence(s, t) -> bool:
    i = 0
    j = 0

    while i < len(s) and j < len(t):

        if s[i] == t[j]:
            i += 1

        j += 1

    return i == len(s)
```

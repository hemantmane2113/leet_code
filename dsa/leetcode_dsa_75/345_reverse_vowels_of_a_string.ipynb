```python
# just reverse the list
def revers_listeVowels_list( s):
    n = len(s)
    left = 0
    right = n - 1

    s_list = list(s)
    print(s_list)

    while left < right:
            s_list[left],s_list[right] = s_list[right],s_list[left]
            left += 1
            right -=1


    return "".join(s_list)
        
```

```python
s_list  = "icecream" #mearceci"

iret = revers_listeVowels_list(s_list)
print(iret)
```

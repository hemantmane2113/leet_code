```python
# brute force solution
#works well but time and space complexity is high
def equalPairs(grid) -> int:

    n = len(grid)
    count = 0


    for i in range(n):
        row = grid[i]
        for j in range(n):
            column = [col[j] for col in grid]
        
            if row == column:
                count += 1

    return count
```

```python
grid = [[3,1,2,2],[1,4,4,5],[2,4,2,2],[2,4,2,2]]

iret = equalPairs(grid)

print(iret)
```

Complexity of the solution

There are n² possible row-column pairs.

1. Loop through every row

code
for i in range(n):

There are n rows.

O(n)

2. For every row, loop through every column

code
for j in range(n):

There are n columns for each row.

So:

n rows × n columns = n² comparisons

O(n²)

3. Creating each column

code:
column = [col[j] for col in grid]

For every column, i go through all n rows to construct the column.

So constructing one column costs:

O(n)

And i construct a column n² times overall because i am doing it inside both loops:

n rows × n columns × n elements

Therefore:

O(n³)

This is actually the biggest reason my solution is O(n³).

4. Comparing row and column

I do:

if row == column:

Comparing two lists can require checking up to n elements.

So one comparison is:

O(n)

And you have n² comparisons:

n² × n = n³

So this is also:

O(n³)

Overall time complexity

We have:

Creating columns → O(n³)
Comparing lists  → O(n³)

Therefore:

Time = O(n³)

We don't add them as O(2n³) because constants are ignored:

O(n³ + n³)
= O(2n³)
= O(n³)
Space complexity

Now look at:

column = [col[j] for col in grid]

At any one point, column contains n elements.

Therefore:

Space = O(n)

I am not storing all columns simultaneously. I am creating one column at a time.

Mine row is just a reference to an existing row:

row = grid[i]

It doesn't create a new n-element list.

So:

column → O(n)
row    → O(1) additional space

Therefore:

Space = O(n)

```python
# best solution
from collections import Counter

def equalPairs(grid) -> int:
    n = len(grid)
    
    # 🔴 CORRECTION 1: Convert inner lists to tuples so they can be hashed
    row_counter = Counter(tuple(row) for row in grid)
    print(row_counter)
    count = 0

    for j in range(n):
        # 🔴 CORRECTION 2: Convert the extracted column to a tuple as well
        column = tuple(grid[row][j] for row in range(n))
        print(column)

        # 🔴 CORRECTION 3: Look up the column directly in row_counter instead of looping
        # This adds the exact number of rows that match this column
        if column in row_counter:
            count += row_counter[column]# fantastic idea !....very nice
            print(count)

    return count
```

```python
grid = [[3,1,2,2],[1,4,4,5],[2,4,2,2],[2,4,2,2]]

iret = equalPairs(grid)

print(iret)
```

Complexity

There are n² elements in the matrix.

Creating the row tuples:

O(n²)

Creating all columns:

O(n²)

Hash-map lookups are O(1) average.

Therefore:

Time: O(n²)

Space: O(n²) because we're storing the rows.

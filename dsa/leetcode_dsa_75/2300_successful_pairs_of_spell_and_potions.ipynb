```python
# works but brute force
# time complexity:O(m×n)-> O(mn) + O(mn) = O(2mn) = O(mn)

def successfulPairs(spells,potions,success):

    prod = []
    result = []
    count = 0

    for s in spells:# O(m)...O(m) for both for loops below
        for p in potions:# O(n)
            prod.append(s * p)
        
        
        for q in prod:# O(n)
            if q >= success:
                count += 1
        
        result.append(count)

        count = 0
        prod = []

    return result
```

```python
# works well,use of binary search
# time complexity: O(nlogn+mlogn)
def successfulPairs(spells,potions,success):
    result = []
 
    sorted_potions = sorted(potions)# can use sort(portions)
    for s in spells:
        req = (success + s - 1) // s

        low = 0
         
        high = len(potions) - 1

        while high >= low:
            mid = low + (high - low)//2

            if sorted_potions[mid] >= req:
                high = mid - 1
            else:
                low = mid + 1

        result.append(len(potions) - low)
    
    return result
```

An Alternative: The Pythonic WayYou can achieve the exact same logic cleanly using Python's built-in bisect library, which handles the binary search logic for you:

```python
import bisect

def successfulPairs(spells, potions, success):
    sorted_potions = sorted(potions)
    n = len(potions)
    result = []
    
    for s in spells:
        req = (success + s - 1) // s
        # Finds the first index where a potion is >= req
        idx = bisect.bisect_left(sorted_potions, req)
        result.append(n - idx)
        
    return result
```

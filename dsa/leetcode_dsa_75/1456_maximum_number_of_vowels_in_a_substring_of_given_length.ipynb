```python
# works but time complexity is high O(n^2)

def maxVowels(s: str, k: int) -> int:
    n = len(s)
    max_count = 0
    count = 0
    vowels = "aeiouAEIOU"
    right = k
    left = 0 

    while left < n:
        for char in s[left:right]:
            if char in vowels:
                count += 1

        max_count = max(count,max_count)

        left += 1
        right += 1
        count = 0

    return max_count
```

```python
# works well
# time complexity O(n)
def maxVowels(s: str, k: int) -> int:
    n = len(s)
    max_count = 0
    count = 0
    vowels = "aeiouAEIOU"
    right = k
    left = 0 

    for i in range(k):
        if s[i] in vowels:
            count += 1

    max_count = max(count,max_count)

    for right in range(k,n):
        if s[left] in vowels:
            count -= 1

        if s[right] in vowels:
            count += 1
        
        max_count = max(count,max_count)

        left += 1
 

    return max_count
```

```python
# works well-best version
# time complexity O(n)
def maxVowels(s: str, k: int) -> int:
    vowels = "aeiou"

    count = 0

    # First window
    for i in range(k):
        if s[i] in vowels:
            count += 1

    max_count = count

    # Slide the window
    for right in range(k, len(s)):
        left = right - k

        if s[left] in vowels:
            count -= 1

        if s[right] in vowels:
            count += 1

        max_count = max(max_count, count)

    return max_count
```

Logic/pattern for fixed size window

1. Build the first window of size k
          ↓
2. Store the window's property/calculate the required information for that window
          ↓
3. Slide the window one position
          ↓
4. Remove the contribution of the element leaving
          ↓
5. Add the contribution of the element entering
          ↓
6. Update the answer
          ↓
7. Repeat

```python
def decodeString(s: str) -> str:
    stack = []
    current = ""
    number = 0

    for char in s:

        if char.isdigit():
            number = number * 10 + int(char)

        elif char == "[":
            stack.append(current)
            stack.append(number)
            current = ""
            number = 0

        elif char.isalpha():
            current += char

        else:  # ']'
            repeat = stack.pop()
            previous = stack.pop()

            current = previous + current * repeat

    return current
```

```python
def decodeString(s: str) -> str:
    stack = []
    current_string = ""
    number = 0

    for char in s:

        if char.isdigit():
            number = number * 10 + int(char)

        elif char == "[":
            stack.append((current_string, number))
            current_string = ""
            number = 0

        elif char.isalpha():
            current_string += char

        elif char == "]":
            previous_string, repeat = stack.pop()
            current_string = previous_string + current_string * repeat

    return current_string
```

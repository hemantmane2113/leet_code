```python
# works well
# time complexity O(log n)
def findPeakElement(nums):
    n = len(nums)

    low = 0
    high = n - 1

    while high > low:# not >= because we're not searching for an exact value.
        mid = low + (high - low) // 2

        if nums[mid] < nums[mid + 1]:# not <=
            low = mid + 1
        else:
            high = mid

    return low # return high
```

```python
nums = [1,2,3,1]
iret = findPeakElement(nums)
print(iret)
```

The entire algorithm is basically this:

             Compare nums[mid] and nums[mid + 1]
                          |
              ┌───────────┴───────────┐
              ↓                       ↓
       nums[mid] < nums[mid+1]   nums[mid] > nums[mid+1]
              ↓                       ↓
          Going UP                 Going DOWN
              ↓                       ↓
        Search RIGHT          Search LEFT or mid
              ↓                       ↓
       low = mid + 1             high = mid

And when:

low == high

only one candidate remains, so:

RETURN low

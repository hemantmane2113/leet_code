```python

def kidsWithCandies(candies, extraCandies):
    greatest = float('-inf')

    result = []

    for i in candies:
        if i > greatest:
            greatest = i

    print(greatest)

    for i in candies:
        print(f"{i} + {extraCandies} = {greatest}")
        if i + extraCandies >= greatest:
            result.append("true")
        else:
            result.append("false")

    
    return result
        
```

```python
candies = [2,3,5,1,3]
extraCandies = 3

kidsWithCandies(candies, extraCandies)
```

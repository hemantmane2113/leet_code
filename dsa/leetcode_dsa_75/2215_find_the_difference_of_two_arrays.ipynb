```python
# best solution

def findDifference(nums1: List[int], nums2: List[int]) -> List[List[int]]:
    # Convert lists to sets to remove duplicates and allow fast math
    set1 = set(nums1)
    set2 = set(nums2)
    
    # set1 - set2 gives elements in set1 that are NOT in set2
    # Convert back to lists at the end
    return [list(set1 - set2), list(set2 - set1)]
```

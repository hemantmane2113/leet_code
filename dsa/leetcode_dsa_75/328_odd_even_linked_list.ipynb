```python
# correct solution 1
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def oddEvenList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        #if not head:
        if not head or not head.next or not head.next.next:     
            return 0
        
        counter = 1
        
        current = head
        previous = None
        
        tail = head
        total_nodes = 1
        while tail.next is not None:
            tail = tail.next
            total_nodes += 1

        while counter <= total_nodes:
            if counter % 2 == 0:
                even_node = current

                previous.next = current.next
                current = current.next


                tail.next = even_node
                even_node.next = None
                tail = even_node
            else:
                previous = current
                current = current.next

            counter += 1

        
        return head
```

```python
# correct solution 2
class Solution:
    def oddEvenList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # Handle empty list or single-node list
        if not head or not head.next:
            return head
        
        odd = head          # Tracks the current odd node
        even = head.next    # Tracks the current even node
        evenHead = even     # Saves the start of the even list to connect later
        
        # Loop while there are still pairs of nodes left to process
        while even and even.next:
            odd.next = even.next    # Link the current odd node to the next odd node
            odd = odd.next          # Move the odd pointer forward
            
            even.next = odd.next    # Link the current even node to the next even node
            even = even.next        # Move the even pointer forward
            
        odd.next = evenHead  # Connect the end of the odd list to the head of the even list
        return head
```

```python
# 1.Helper function to turn a Python list into Linked List nodes
def make_linked_list(arr):
    if not arr: return None
    
    head = ListNode(arr[0])
    current = head
    for num in arr[1:]:
        current.next = ListNode(num)
        current = current.next
    return head
```

```python
# 2.Helper function Convert Linked List back to Python list for easy printing
def print_linked_list(head):
    result = []
    current = head
    while current:
        result.append(str(current.val))
        current = current.next
    print(" -> ".join(result))
```

```python
#best solution - internet
def minEatingSpeed(piles,h):
  
    low = 1
    high = max(piles)

    while high >= low:
        k = low + (high - low) // 2

        hours = 0

        for pile in piles:
            hours += (pile + k - 1)//k # means: How many groups of size k are needed to cover pile bananas?

        if hours <= h:
            high = k - 1
        else:
            low =  k + 1

    return low
```

```python
piles = [3,6,7,11]
h = 8
iret =  minEatingSpeed(piles,h)
print(iret)
```

The formula (pile + k - 1) // k is probably the most confusing part of Koko's problem. Let's understand why we use:

(pile + k - 1) // k

instead of just:
pile // k

1. What are we trying to calculate?

For each pile, we want to know:

How many hours does Koko need to finish this pile if she eats k bananas per hour?

Suppose:

pile = 7
k = 3

Koko eats 3 bananas per hour.

Hour 1: 7 → 4
Hour 2: 4 → 1
Hour 3: 1 → 0

So she needs:

3 hours

Mathematically:

7 / 3 = 2.333...

We need to round UP:

ceil(7 / 3) = 3

2. Why can't we use pile // k?

Python:

7 // 3

gives:

2

But that's wrong for our problem.

Koko can't eat for 2.33 hours.

She needs the third hour to finish the remaining banana.

So we need:

ceil(7 / 3) = 3

3. How do we get ceiling using integer division?

Here's the trick:

(pile + k - 1) // k

For:

pile = 7
k = 3

we get:

(7 + 3 - 1) // 3

which is:

9 // 3

which gives:

3

Exactly what we want.

4. Why does adding k - 1 work?

Let's look at several examples.

Example 1
pile = 6
k = 3

Exactly divisible:

6 / 3 = 2

Formula:

(6 + 3 - 1) // 3
= 8 // 3
= 2

Good.

Example 2
pile = 7
k = 3
7 / 3 = 2.33

We need 3.

Formula:

(7 + 3 - 1) // 3
= 9 // 3
= 3

Good.

Example 3
pile = 8
k = 3
8 / 3 = 2.66

We need 3.

Formula:

(8 + 3 - 1) // 3
= 10 // 3
= 3

Good.

Example 4
pile = 10
k = 3
10 / 3 = 3.33

We need 4.

Formula:

(10 + 3 - 1) // 3
= 12 // 3
= 4

Good.

5. Think about it without mathematics

Suppose Koko eats:

k = 3 bananas/hour

And there are:

7 bananas

Divide them into groups of 3:

🍌🍌🍌 | 🍌🍌🍌 | 🍌
  ↑          ↑       ↑
Hour 1     Hour 2   Hour 3

There are 2 complete groups plus some bananas left over.

That leftover requires another hour.

So:

7 bananas / 3 per hour = 3 hours

The formula:

(pile + k - 1) // k

is simply a clever way of doing:

integer division, but round UP instead of down.

6. And that's why code has this:

hours = 0

for pile in piles:
    hours += (pile + k - 1) // k

For:

piles = [3, 6, 7, 11]
k = 4

we calculate:

3  → (3 + 4 - 1) // 4 = 6 // 4  = 1
6  → (6 + 4 - 1) // 4 = 9 // 4  = 2
7  → (7 + 4 - 1) // 4 = 10 // 4 = 2
11 → (11 + 4 - 1) // 4 = 14 // 4 = 3

Then:

hours = 1 + 2 + 2 + 3
      = 8

So Koko needs 8 hours at speed k = 4.

The one thing to remember
(pile + k - 1) // k

means:

How many groups of size k are needed to cover pile bananas?

Or simply:

ceil(pile / k) without using floating-point division.

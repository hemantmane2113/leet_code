```python
# the best code

def findMaxAverage(nums: List[int], k: int) -> float:
    left = 0
    ele_sum = 0

    # First window
    for i in range(k):
        ele_sum += nums[i]

    max_avg = ele_sum / k

    # Slide the window
    for right in range(k, len(nums)):
        ele_sum += nums[right] - nums[left]
        left += 1

        max_avg = max(max_avg, ele_sum / k)

    return max_avg
```

Logic/pattern for fixed size window

1. Build the first window of size k
          ↓
2. Store the window's property/calculate the required information for that window
          ↓
3. Slide the window one position
          ↓
4. Remove the contribution of the element leaving
          ↓
5. Add the contribution of the element entering
          ↓
6. Update the answer
          ↓
7. Repeat

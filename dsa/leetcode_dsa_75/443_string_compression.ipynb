```python
def compress(chars):
    n = len(chars)

    count = 0
    left = 0
    right = 0
    write = 0

    while right < n:

        if chars[left] == chars[right]:
            count += 1
            right += 1

        else:
            # Write the character
            chars[write] = chars[left]
            write += 1

            # Write the count only if > 1
            if count > 1:
                for digit in str(count):
                    chars[write] = digit
                    write += 1

            # Move to the next group
            left = right
            count = 0

    # Process the last group
    chars[write] = chars[left]
    write += 1

    if count > 1:
        for digit in str(count):
            chars[write] = digit
            write += 1
  
    return write
```

```python
def compress(chars):
    write = 0
    read = 0

    while read < len(chars):

        # Find the end of the current group
        start = read

        while read < len(chars) and chars[read] == chars[start]:
            read += 1

        # Write the character
        chars[write] = chars[start]
        write += 1

        # Calculate frequency
        count = read - start

        # Write frequency if > 1
        if count > 1:
            for digit in str(count):
                chars[write] = digit
                write += 1

    return write
```

```python
class RecentCounter:

    def __init__(self):
        self.my_queue = []
        
    def ping(self, t: int) -> int:

        self.my_queue.append(t)

        while self.my_queue[0] < t - 3000:
            self.my_queue.pop(0)
        
        return len(self.my_queue)
```

```python
obj1 = RecentCounter()
```

```python
obj1.ping(1)
obj1.ping(100)
obj1.ping(3001)
obj1.ping(3002)
```

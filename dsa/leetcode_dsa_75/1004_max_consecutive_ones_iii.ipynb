Logic/pattern for variable size window

1. Start with an empty window
          ↓
2. Expand the window by moving `right`
          ↓
3. Update the window's property/state
          ↓
4. Check whether the window is valid
          ↓
5. If the window is invalid, move `left`
   to shrink the window
          ↓
6. While shrinking, remove the contribution
   of the element leaving
          ↓
7. Keep shrinking until the window becomes valid
          ↓
8. Calculate the current window's result
          ↓
9. Update the answer
          ↓
10. Repeat by expanding `right`

```python
def maxArea(height) -> int:

    n = len(height)

    left = 0

    right = n - 1

    max_container_volume = 0

    while (left < n) and  (right >= 0):

        container_height = min(height[left],height[right])

        width = right - left


        container_volume = width * container_height

        max_container_volume = max(max_container_volume,container_volume)


        if height[left] < height[right]:
            left += 1
        else:
            right -= 1


    return max_container_volume
```

```python
def maxArea(height) -> int:

    left = 0
    right = len(height) - 1

    max_container_volume = 0

    while left < right:

        container_height = min(height[left], height[right])
        width = right - left

        container_volume = width * container_height

        max_container_volume = max(max_container_volume,container_volume)

        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    return max_container_volume
```

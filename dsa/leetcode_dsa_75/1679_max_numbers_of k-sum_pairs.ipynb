```python
# correct but some non useful things present
from collections import Counter
def maxOperations(nums: List[int], k: int) -> int:

    n = len(nums)
    left = 0
    right = 0
    count = 0

    complimentary_num = 0 
    seen_num_list = []

    seen = Counter()


    while left < n :
        complimentary_num =  k - nums[left] 
   
        
        if complimentary_num in seen_num_list and seen[complimentary_num] >= 1:
            count += 1
            seen[complimentary_num] -= 1
            
            
        else:
            seen_num_list.append(nums[left])
            seen[nums[left]] += 1
            
               
        left += 1
       
    return count 
```

```python
# using hash map method/counter module # time complexity: O(n)
from collections import Counter
def maxOperations(nums: List[int], k: int) -> int:

    n = len(nums)
    left = 0
    count = 0

    complimentary_num = 0 
    
    seen = Counter()


    while left < n :
        complimentary_num =  k - nums[left] 
   
        
        if  seen[complimentary_num] >= 1:
            count += 1
            seen[complimentary_num] -= 1
            
            
        else:
            
            seen[nums[left]] += 1
            
               
        left += 1
       
    return count 
```

```python
# using 2 pointer method ,time complexity:O(n log n)
# need to sort the array first
def maxOperations(nums, k): 
    nums.sort()

    left = 0
    right = len(nums) - 1
    count = 0

    while left < right:

        current_sum = nums[left] + nums[right]

        if current_sum == k:
            count += 1
            left += 1
            right -= 1

        elif current_sum < k:
            left += 1

        else:
            right -= 1

    return count
```

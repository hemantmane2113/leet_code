```python
def reverseWords(s: str) -> str:
    
    string_list = s.split(" ")
    print(string_list)
    n = len(string_list)

    reversed_str = ""

    for i in range(n-1,-1,-1):
        reversed_str = reversed_str + " "+ string_list[i]
    print(reversed_str)

    return reversed_str
```

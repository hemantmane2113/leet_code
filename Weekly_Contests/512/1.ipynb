```python
def largestInteger(self, n: int, s: int) -> int:
    # If the required sum is impossible to achieve with n digits
    if s < 0 or s > 9 * n:
        return -1
    # Edge case: If sum is 0, it's only possible if it's a 1-digit number "0"
    if s == 0:
        return 0 

    result = []
    
    for i in range(n):
        # Place the largest possible digit (up to 9) that leaves enough 
        # sum remaining for the rest of the digits
        digit = min(9, s)
        result.append(str(digit))
        s -= digit
        
    return int("".join(result))
```

```python
n = 2
s = 19
iret = largestInteger(None,n,s)
print(iret)
```

```python
from datetime import datetime
def secondsBetweenTimes(startTime,endTime):
    time_format = "%H:%M:%S"

    start = datetime.strptime(startTime,time_format)
    end = datetime.strptime(endTime,time_format)
    time_diff =  end - start

    return int(time_diff.total_seconds())
```

```python
startTime = "12:34:56"
endTime = "13:00:00"
iret = secondsBetweenTimes(startTime,endTime)
print(iret)
```

```python
def minimumCost(self, nums: list[int], k: int) -> int:
    n = len(nums)
    left_over = k
    min_cost = 0

    i = 0
    while i < n:
        diff = nums[i] - left_over
        if diff > 0:
            hops = (diff + k - 1) // k
            left_over += hops * k
            min_cost += hops 

        left_over -= nums[i]
        i += 1
    return min_cost
```

```python

def minimumCost(self, nums: list[int], k: int) -> int:
    n = len(nums)
    min_cost = 0
    cur_capacity = k  # Tracks our current number line capacity threshold
    MOD = 10**9 + 7   # Standard LeetCode modulo for large numbers

    i = 0
    while i < n:
        # Check how much the current number exceeds our capacity threshold
        diff = nums[i] - cur_capacity
        
        if diff > 0:
            # Math shortcut to see how many hops of size 'k' we need
            hops = (diff + k - 1) // k
            cur_capacity += hops * k
            min_cost += hops  # Tracks total jumps made
            
        # Consume the current element from our capacity line
        cur_capacity -= nums[i]
        i += 1  # Fixed: Always increment i to prevent infinite loops!
        
    # Apply your number line arithmetic progression formula: 1 + 2 + 3 + 4...
    min_cost %= MOD
    return (1 + min_cost) * min_cost // 2 % MOD
```

```python
# best solution
def minimumCost(self, nums: list[int], k: int) -> int:
    cnt = 0
    cur = k
    MOD = 10**9 + 7
    
    for x in nums:
        # Check how much the element exceeds our current line/capacity
        diff = x - cur
        if diff > 0:
            # Math formula to calculate how many times we need to increment 'k'
            m = (diff + k - 1) // k
            cur += m * k
            cnt += m  # 'cnt' increments like 1, 2, 3...
        
        cur -= x
        
    # The return value sums the sequence (1 + 2 + 3 + ...) using arithmetic series formula
    cnt %= MOD
    return (1 + cnt) * cnt // 2 % MOD
```

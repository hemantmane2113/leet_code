```python
list1 = [[1,6],[2,5],[3,7]]

for listt in list1:
    print(min(sum(listt)))  
```

```python
time = [sum(listt) for listt in list1]
print(time)
print(min(time))
```

```python
tasks = [[1,8],[2,5],[3,7]]
for s,t in tasks:# will work as the number of elements in the sublists are the same
     print(s,t)
```

```python
tasks = [[1,8],[2,5],[3,7,8]]
for s,t in tasks:# will not work as the number of elements in the sublists are not the same
     print(s,t)
```

```python
tasks = [[1,8],[2,5],[3,7,8]]
for task in tasks:
     for s,t in task:
         print(s,t)
```

```python
tasks = [[1,8],[2,5],[3,7,8]]
for task in tasks:
     for items in task:
         print(items)
```

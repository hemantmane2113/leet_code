```python
# 1. Using standard decimal integers
decimal_result = 4 | 2
print(decimal_result)       # Output: 6

# 2. Doing the exact same thing using binary literals (0b prefix)
binary_result = 0b0100 | 0b0010
print(binary_result)        # Output: 6 (Python still prints it in decimal!)

# 3. Viewing the result back in binary format
print(bin(decimal_result))  # Output: '0b110'
```
